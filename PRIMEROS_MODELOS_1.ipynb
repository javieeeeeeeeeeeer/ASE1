```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

# Datos ficticios de países en los Juegos Olímpicos
data = pd.DataFrame({
    'Poblacion': [10, 50, 100, 200, 300],  # en millones
    'PIB': [1000, 5000, 15000, 30000, 50000],  # en millones de dólares
    'Mortalidad_Infantil': [50, 30, 20, 10, 5],  # por cada 1000 nacidos
    'Medallas': [2, 10, 25, 40, 60]  # Medallas ganadas
})

# Definir variables independientes y dependiente
X = data[['Poblacion', 'PIB', 'Mortalidad_Infantil']]
y = data['Medallas']

# Agregar una constante para la intersección
X = sm.add_constant(X)

# Ajustar el modelo de regresión lineal múltiple
modelo = sm.OLS(y, X).fit()

# Ver resultados
print(modelo.summary())
```

```python
data['Proporcion_Medallas'] = data['Medallas'] / data['Medallas'].sum()
```

```python
from statsmodels.discrete.discrete_model import MNLogit

# Definir variables dependiente e independiente
X = data[['Poblacion', 'PIB', 'Mortalidad_Infantil']]
y = data['Proporcion_Medallas']

# Agregar una constante
X = sm.add_constant(X)

# Ajustar modelo logístico multinomial
modelo_logit = MNLogit(y, X).fit()

# Ver resultados
print(modelo_logit.summary())
```
