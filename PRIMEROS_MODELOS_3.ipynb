```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

# Definir los datos
data = {
    "Height": [160, 170, 180, 190, 200],
    "Weight": [55, 65, 72, 80, 90]
}

df = pd.DataFrame(data)

# Definir las variables X (altura) e Y (peso)
X = df["Height"]
Y = df["Weight"]

# Agregar una columna de unos para calcular B0
X = sm.add_constant(X)

# Crear el modelo de regresión
model = sm.OLS(Y, X).fit()

# Mostrar los coeficientes
print(model.summary())
```
