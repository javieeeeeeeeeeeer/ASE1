## Data preparation

### Subtask:
Rename the "Country Name" column to "Country" in both `medallas_df` and `cd_df` DataFrames.


**Reasoning**:
Rename the "Country Name" column to "Country" in both DataFrames.



```python
medallas_df = medallas_df.rename(columns={"Country Name": "Country"})
cd_df = cd_df.rename(columns={"Country Name": "Country"})
```

## Data wrangling

### Subtask:
Perform a left join between `cd_df` and `medallas_df` using the "Country" column as the key.


**Reasoning**:
Perform a left join between `cd_df` and `medallas_df` using the "Country" column as the key.



```python
merged_df = cd_df.merge(medallas_df, on='Country', how='left')
```

## Data cleaning

### Subtask:
Fill missing values in the "Total Medals" and "Gold Medals" columns of `merged_df` with 0.


**Reasoning**:
Fill the missing values in the "Total Medals" and "Gold Medals" columns with 0.



```python
merged_df['total_medals'] = merged_df['total_medals'].fillna(0)
merged_df['total_gold_medals'] = merged_df['total_gold_medals'].fillna(0)
merged_df.to_excel('TABLAFINAL.xlsx', index=False)
```
