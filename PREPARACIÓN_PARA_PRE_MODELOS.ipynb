```python
# Asegurarnos de que la columna es numérica
df["PIB_POR_PAIS"] = pd.to_numeric(df["PIB_POR_PAIS"], errors="coerce")

# Diagnóstico: cuántos valores faltan
missing_pib = df["PIB_POR_PAIS"].isna().sum()
total_rows = len(df)
print(f"PIB: {missing_pib} valores faltantes de {total_rows} filas totales.")
```

```python
sedes_olimpicas = {
    1960: "Italy",
    1964: "Japan",
    1968: "Mexico",
    1972: "Germany",
    1976: "Canada",
    1980: "Russia",  # USSR
    1984: "United States",
    1988: "South Korea",
    1992: "Spain",
    1996: "United States",
    2000: "Australia",
    2004: "Greece",
    2008: "China",
    2012: "United Kingdom",
    2016: "Brazil",
    2020: "Japan",
    2024: "France"
}
# Crear la variable LOCAL con base en el diccionario de sedes
df["LOCAL"] = df.apply(lambda row: 1 if sedes_olimpicas.get(row["YEAR"]) == row["COUNTRY"] else 0, axis=1)
```

```python
# Mostrar los países marcados como anfitriones
df[df["LOCAL"] == 1][["YEAR", "COUNTRY", "LOCAL"]].sort_values("YEAR")
```

```python
vecinos_por_sede = {
    1960: ["France", "Switzerland", "Austria", "Germany", "Slovenia"],  # Italy
    1964: ["South Korea", "China", "Russia"],  # Japan (isla, pero geopolíticamente cercanos)
    1968: ["United States", "Guatemala", "Belize"],  # Mexico
    1972: ["France", "Austria", "Poland", "Switzerland", "Netherlands", "Czech Republic"],  # Germany
    1976: ["United States"],  # Canada
    1980: ["China", "Mongolia", "Ukraine", "Belarus", "Kazakhstan", "Poland"],  # USSR (simplificado como Russia)
    1984: ["Canada", "Mexico"],  # USA
    1988: ["China", "Japan", "North Korea"],  # South Korea
    1992: ["Portugal", "France", "Italy"],  # Spain
    1996: ["Canada", "Mexico"],  # USA
    2000: ["New Zealand"],  # Australia
    2004: ["Albania", "Bulgaria", "North Macedonia", "Turkey"],  # Greece
    2008: ["Russia", "Vietnam", "India", "Mongolia", "Pakistan", "Kazakhstan", "North Korea"],  # China
    2012: ["France", "Germany", "Ireland", "Netherlands", "Belgium"],  # UK
    2016: ["Argentina", "Uruguay", "Paraguay", "Bolivia", "Peru"],  # Brazil
    2020: ["South Korea", "China", "Russia"],  # Japan
    2024: ["Germany", "Belgium", "Italy", "Spain", "Switzerland"]  # France
}
def es_vecino(row):
    vecinos = vecinos_por_sede.get(row["YEAR"], [])
    return 1 if row["COUNTRY"] in vecinos else 0

df["VECINO"] = df.apply(es_vecino, axis=1)

```

```python
# Mostrar los países marcados como anfitriones
df[df["VECINO"] == 1][["YEAR", "COUNTRY", "VECINO"]].sort_values("YEAR")
```

```python
# Aproximación de índice de precios por año (base 2020 = 100)
# Cuanto más viejo el año, más alto el ajuste
indice_precios = {
    1960: 0.10, 1964: 0.13, 1968: 0.16, 1972: 0.20, 1976: 0.25,
    1980: 0.30, 1984: 0.35, 1988: 0.40, 1992: 0.50, 1996: 0.60,
    2000: 0.70, 2004: 0.80, 2008: 0.90, 2012: 0.95, 2016: 0.98,
    2020: 1.00, 2024: 1.03
}

# Crear una nueva columna con el índice de precios correspondiente
df["INDICE_PRECIOS"] = df["YEAR"].map(indice_precios)

# Calcular PIB real
df["PIB_REAL"] = df["PIB_POR_PAIS"] / df["INDICE_PRECIOS"]

# Calcular logaritmo del PIB real (evitamos log de 0 o NaN)
import numpy as np
df["LOG_PIB_REAL"] = np.log(df["PIB_REAL"].replace(0, np.nan))
```

```python
# Calcular el total de medallas entregadas por año
total_medallas_por_ano = df.groupby("YEAR")["MEDALLAS_TOTALES_POR_PAIS_POR_AÑO"].transform("sum")

# Calcular la proporción de medallas por país y año
df["PROP_MEDALLAS"] = df["MEDALLAS_TOTALES_POR_PAIS_POR_AÑO"] / total_medallas_por_ano

df["LOG_PROP_MEDALLAS"] = np.log(df["PROP_MEDALLAS"].replace(0, np.nan))

```

```python
# Mostrar países con al menos una medalla
df[df["MEDALLAS_TOTALES_POR_PAIS_POR_AÑO"] > 0][[
    "COUNTRY",
    "YEAR",
    "MEDALLAS_TOTALES_POR_PAIS_POR_AÑO",
    "PROP_MEDALLAS",
    "LOG_PROP_MEDALLAS"
]].sample(15, random_state=1)  # Puedes cambiar el número o quitar random_state
```
