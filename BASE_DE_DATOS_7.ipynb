## Data loading

### Subtask:
Load the two CSV files into pandas DataFrames.


**Reasoning**:
Load the two CSV files into pandas DataFrames, handling potential encoding issues.



```python
import pandas as pd

try:
    df_final = pd.read_csv('TABLAFINALCONLAQUEPUEDOHACERCOSAS.csv', encoding='latin-1')
except UnicodeDecodeError:
    try:
        df_final = pd.read_csv('TABLAFINALCONLAQUEPUEDOHACERCOSAS.csv', encoding='utf-8')
    except Exception as e:
        print(f"Error loading TABLAFINALCONLAQUEPUEDOHACERCOSAS.csv: {e}")
        df_final = None

try:
    df_pib = pd.read_csv('PIB BUENO4.csv', encoding='latin-1')
except UnicodeDecodeError:
    try:
        df_pib = pd.read_csv('PIB BUENO4.csv', encoding='utf-8')
    except Exception as e:
        print(f"Error loading PIB BUENO4.csv: {e}")
        df_pib = None

if df_final is not None:
    display(df_final.head())
if df_pib is not None:
    display(df_pib.head())
```

## Data exploration

### Subtask:
Explore the loaded dataframes `df_final` and `df_pib` to understand their structure and identify potential issues.


**Reasoning**:
Examine `df_final` and `df_pib` to understand their structure, identify missing values, and summarize key characteristics of relevant columns, as per the instructions.



```python
# Examine df_final
print("Shape of df_final:", df_final.shape)
display(df_final.head())
print("\nData types of df_final columns:\n", df_final.dtypes)
print("\nMissing values in df_final:\n", df_final.isnull().sum())

print("\nSummary of 'Country Name 2' column in df_final:")
print("Data type:", df_final['Country Name 2'].dtype)
print("Number of missing values:", df_final['Country Name 2'].isnull().sum())
print("Unique values:", df_final['Country Name 2'].nunique())
print("Most frequent values:\n", df_final['Country Name 2'].value_counts().head())


# Examine df_pib
print("\n\nShape of df_pib:", df_pib.shape)
display(df_pib.head())
print("\nData types of df_pib columns:\n", df_pib.dtypes)
print("\nMissing values in df_pib:\n", df_pib.isnull().sum())

print("\nSummary of 'Country name' column in df_pib:")
print("Data type:", df_pib['Country Name'].dtype)
print("Number of missing values:", df_pib['Country Name'].isnull().sum())
print("Unique values:", df_pib['Country Name'].nunique())
print("Most frequent values:\n", df_pib['Country Name'].value_counts().head())

print("\nSummary of 'PIB PROMEDIO POR PAÍS' column in df_pib:")
print("Data type:", df_pib['PIB PROMEDIO POR PAÍS'].dtype)
print("Number of missing values:", df_pib['PIB PROMEDIO POR PAÍS'].isnull().sum())
print("Descriptive statistics:\n", df_pib['PIB PROMEDIO POR PAÍS'].describe())
```

## Data preparation

### Subtask:
Prepare the data for merging by renaming a column in `df_pib` and handling missing values in the join key column ("Country Name 2") of both DataFrames.


**Reasoning**:
Rename the 'Country Name' column in `df_pib` to 'Country Name 2' and handle the missing values in the 'Country Name 2' column of both dataframes. Since there are only 2 missing values in 'Country Name' column of `df_pib`, and no missing values in 'Country Name 2' column of `df_final`, dropping those two rows in `df_pib` seems a reasonable choice.



```python
# Rename the 'Country Name' column in df_pib
df_pib = df_pib.rename(columns={'Country Name': 'Country Name 2'})

# Drop rows with missing values in 'Country Name 2' column of df_pib
df_pib_cleaned = df_pib.dropna(subset=['Country Name 2'])

# Verify the changes
print("Number of missing values in 'Country Name 2' column of df_final:", df_final['Country Name 2'].isnull().sum())
print("Number of missing values in 'Country Name 2' column of df_pib:", df_pib_cleaned['Country Name 2'].isnull().sum())
```

## Data wrangling

### Subtask:
Merge the two dataframes, `df_final` and `df_pib_cleaned`, based on the common column "Country Name 2".


**Reasoning**:
Merge the two dataframes based on the "Country Name 2" column using an inner join, then check for and remove duplicates.



```python
# Merge the two dataframes
df_merged = pd.merge(df_final, df_pib_cleaned, on='Country Name 2', how='inner')

# Check the shape of the merged dataframe
print("Shape of df_merged:", df_merged.shape)
display(df_merged.head())

# Remove duplicate rows
df_merged = df_merged.drop_duplicates(keep='first')

# Display the updated dataframe
print("\nShape of df_merged after removing duplicates:", df_merged.shape)
display(df_merged.head())
```
